```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Data
df = pd.read_csv("survey-data.csv")

# 2. Task 1.1: Top 5 Databases Respondents Want to Work With (Jawaban Pertanyaan #1)
top_5_db = df['DatabaseWantToWorkWith'].value_counts().head(5)

# 3. Visualisasi Pie Chart
plt.figure(figsize=(8, 8))
plt.pie(top_5_db, labels=top_5_db.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0'])
plt.title('Top 5 Databases Respondents Want to Work With')
plt.show()

# 4. Tutup Koneksi (Jawaban Pertanyaan #2)
# Walaupun pakai Pandas, jalankan ini untuk formalitas kuis
print("SQLite connection closed.")
```
