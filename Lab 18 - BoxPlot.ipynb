```python
# Langkah A: Download database asli (Wajib di notebook baru)
!wget https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/QR9YeprUYhOoLafzlLspAw/survey-results-public.sqlite

import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Langkah B: Connect ke file yang baru di-download
conn = sqlite3.connect('survey-results-public.sqlite')

# Langkah C: Cek apakah tabel 'main' sudah ada
QUERY_CEK = "SELECT name FROM sqlite_master WHERE type='table'"
print("Daftar tabel yang tersedia:", pd.read_sql_query(QUERY_CEK, conn))

# Langkah D: Jika 'main' sudah muncul, jalankan box plot
QUERY_AGE = "SELECT Age FROM main"
df_age = pd.read_sql_query(QUERY_AGE, conn)
df_age['Age_Numeric'] = df_age['Age'].str.extract('(\d+)').astype(float)

plt.figure(figsize=(8,6))
sns.boxplot(x=df_age['Age_Numeric'], color='lightgreen')
plt.title('Box Plot Umur Responden')
plt.show()
```
