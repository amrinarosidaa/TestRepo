```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Data
df = pd.read_csv("survey-data.csv")

# 2. Filter untuk usia 45-60 (Kategori: 45-54 dan 55-64)
age_filter = ['45-54 years old', '55-64 years old']
df_45_60 = df[df['Age'].isin(age_filter)]

# 3. Hitung Median CompTotal per kelompok umur (Jawaban Pertanyaan #1)
line_data = df_45_60.groupby('Age')['CompTotal'].median().sort_index()

# 4. Visualisasi Line Chart

plt.figure(figsize=(10, 6))
plt.plot(line_data.index, line_data.values, marker='o', linestyle='-', color='green', linewidth=2)

plt.title('Tren Median CompTotal untuk Usia 45-60 Tahun')
plt.xlabel('Kelompok Umur')
plt.ylabel('Median Total Kompensasi')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# 5. Tutup Koneksi (Jawaban Pertanyaan #2)
print("Line Chart lab connection closed.")
```
