```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Data
file_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(file_url)

# 2. Pembersihan Data (Pastikan mengambil angka)
df['TimeSearching_Num'] = df['TimeSearching'].astype(str).str.extract('(\d+)').astype(float)
df['TimeAnswering_Num'] = df['TimeAnswering'].astype(str).str.extract('(\d+)').astype(float)

# 3. Filter Umur (Cari yang mengandung angka '30' dan '34')
# Cara ini lebih aman daripada menulis teks lengkapnya
df_30_35 = df[df['Age'].astype(str).str.contains('30-34')]

# CEK DATA: Jika ini 0, berarti filternya gagal
print(f"Jumlah data untuk umur 30-34: {len(df_30_35)} baris")

# 4. Hitung Median & Isi data kosong dengan 0 agar batang muncul
median_data = df_30_35[['TimeSearching_Num', 'TimeAnswering_Num']].median().fillna(0)

# 5. Visualisasi Stacked Bar Chart
if len(df_30_35) > 0:
    plot_data = pd.DataFrame(median_data).T
    
    # Gunakan ax agar kontrol grafik lebih stabil
    ax = plot_data.plot(kind='bar', stacked=True, color=['skyblue', 'salmon'], figsize=(8,6))

    plt.title('Median Waktu Aktivitas (Searching vs Answering) - Umur 30-34')
    plt.xlabel('Grup Umur 30-34 Tahun')
    plt.ylabel('Menit (Median)')
    plt.xticks([]) 
    plt.legend(['Waktu Mencari Solusi', 'Waktu Menjawab'])
    
    # Tambahkan label angka di atas batang (opsional agar makin keren)
    for p in ax.patches:
        width, height = p.get_width(), p.get_height()
        x, y = p.get_xy() 
        if height > 0:
            ax.annotate(f'{height:.0f} min', (x + width/2, y + height/2), ha='center', va='center', fontsize=12, color='white', fontweight='bold')
            
    plt.show()
else:
    print("Grafik tidak muncul karena data umur 30-34 tidak ditemukan. Cek kembali kolom 'Age' kamu!")
```
