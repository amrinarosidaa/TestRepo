```python
# 1. Histogram CompTotal (Tugas 1.1)
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

# Pastikan koneksi ke database sudah dibuat kembali
conn = sqlite3.connect('survey-data.sqlite') 

# Baris kode yang tadi error sekarang akan berjalan lancar:
QUERY = "SELECT CompTotal FROM main"
df_comp = pd.read_sql_query(QUERY, conn)

# Visualisasi Histogram
plt.figure(figsize=(10, 6))
plt.hist(df_comp['CompTotal'].dropna(), bins=50, color='skyblue', edgecolor='black')
plt.title('Distribusi Total Kompensasi')
plt.show()
```
